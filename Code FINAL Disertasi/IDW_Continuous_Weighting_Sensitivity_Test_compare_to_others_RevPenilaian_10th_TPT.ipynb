```python
#1) Load data & matrices

import numpy as np
import pandas as pd
import scipy.sparse as sp
from scipy.sparse import csr_matrix, kron
import libpysal
from spreg import ML_Lag, GM_Lag

DATA_FILE   = "data_JarakCentroid_10th.xlsx"
MATRIX_FILE = "matrices_JarakCentroid_10th.xlsx"

df = pd.read_excel(DATA_FILE)
xls = pd.ExcelFile(MATRIX_FILE)
print("Sheets:", xls.sheet_names)

def read_mat(sheet):
    return pd.read_excel(MATRIX_FILE, sheet_name=sheet, header=None).to_numpy(dtype=float)

# shared matrices
W_T = read_mat("W_T")  # 10x10
I_T = read_mat("I_T")  # 10x10
I_S = read_mat("I_S")  # 35x35

# distance matrix (required)
D   = read_mat("D")    # 35x35

assert df.shape[0] == 350
assert W_T.shape == (10,10)
assert I_T.shape == (10,10)
assert I_S.shape == (35,35)
assert D.shape   == (35,35)

n, T = 35, 10
```

```python
#2) Build Y matrix (Tn) sesuai urutan stacking (blok per tahun)

years_order = list(pd.unique(df["Waktu"]))
print("years_order:", years_order)
assert len(years_order) == 10

y_vec = df["Y"].astype(float).to_numpy()
Y = y_vec.reshape(T, n)
print("Y shape:", Y.shape)
```

```python
#3) Build W_S^ID: Inverse Distance + row-normalization

def build_WS_inverse_distance(D, power=1.0, eps=1e-6, use_cutoff=False, cutoff=None):
    """
    D: (35x35) distance matrix
    power: p in 1/(d^p + eps)
    use_cutoff: if True, set weights to 0 for D > cutoff
    cutoff: distance threshold (same unit as D)
    """
    D = np.array(D, dtype=float, copy=True)
    n = D.shape[0]

    WS = np.zeros((n, n), dtype=float)
    for i in range(n):
        for j in range(n):
            if i == j:
                WS[i, j] = 0.0
            else:
                WS[i, j] = 1.0 / ((D[i, j] ** power) + eps)

    if use_cutoff:
        if cutoff is None:
            raise ValueError("cutoff must be provided if use_cutoff=True")
        WS[D > cutoff] = 0.0
        np.fill_diagonal(WS, 0.0)

    # row-normalization
    rs = WS.sum(axis=1, keepdims=True)
    WS = np.divide(WS, rs, out=np.zeros_like(WS), where=rs != 0)

    return WS

# --- choose settings ---
W_S_idw = build_WS_inverse_distance(D, power=1.0, eps=1e-6, use_cutoff=False)

print("W_S_idw:", W_S_idw.shape, "row sum min/max:", W_S_idw.sum(1).min(), W_S_idw.sum(1).max())
```

```python
#4) Build W_ST^ID via Kronecker Eq. (10)

def build_WST(W_S, W_T, I_S, I_T):
    W_ST_sp = kron(csr_matrix(I_T), csr_matrix(W_S), format="csr") + \
              kron(csr_matrix(W_T), csr_matrix(I_S), format="csr")
    W_ST = W_ST_sp.toarray()
    np.fill_diagonal(W_ST, 0.0)
    return W_ST

W_ST_idw = build_WST(W_S_idw, W_T, I_S, I_T)
print("W_ST_idw:", W_ST_idw.shape)
```

```python
#OUTPUT 1: Moran-2Di inference (within-year permutation)

def moran_2di(z_vec, W_ST):
    z = np.asarray(z_vec, dtype=float).reshape(-1, 1)
    N = z.shape[0]
    S0 = float(np.sum(W_ST))
    if S0 == 0:
        raise ValueError("S0 = sum(W_ST) is zero.")
    num = float(z.T @ W_ST @ z)
    den = float(z.T @ z)
    return (N / S0) * (num / den)

def mc_within_year_permutation_moran2di(z, years, W_ST, B=1000, seed=2026, two_sided=True):
    rng = np.random.default_rng(seed)
    z = np.asarray(z, dtype=float).ravel()
    years = np.asarray(years).ravel()

    I_obs = moran_2di(z, W_ST)

    uniq = np.unique(years)
    idxs = {yr: np.where(years == yr)[0] for yr in uniq}

    I_perm = np.empty(B, dtype=float)
    for b in range(B):
        z_perm = z.copy()
        for yr, idx in idxs.items():
            z_perm[idx] = rng.permutation(z_perm[idx])
        I_perm[b] = moran_2di(z_perm, W_ST)

    I_mean = float(I_perm.mean())
    ci_low, ci_high = float(np.quantile(I_perm, 0.025)), float(np.quantile(I_perm, 0.975))

    if two_sided:
        p_value = (np.sum(np.abs(I_perm) >= np.abs(I_obs)) + 1) / (B + 1)
    else:
        p_value = (np.sum(I_perm >= I_obs) + 1) / (B + 1)

    return {"I_obs": float(I_obs), "I_mean": I_mean, "p_value": float(p_value),
            "CI95_low": ci_low, "CI95_high": ci_high}

# prepare z and years
y_all = df["Y"].astype(float).to_numpy()
z = (y_all - y_all.mean()) / y_all.std(ddof=0)
years_vec = df["Waktu"].to_numpy()

out_moran_idw = mc_within_year_permutation_moran2di(z, years_vec, W_ST_idw, B=1000, seed=2026, two_sided=True)

print("=== Moran-2Di (Inverse Distance baseline) ===")
print("I_obs   :", out_moran_idw["I_obs"])
print("I_mean  :", out_moran_idw["I_mean"])
print("p-value :", out_moran_idw["p_value"])
print("CI95    :", (out_moran_idw["CI95_low"], out_moran_idw["CI95_high"]))
```

```python
#OUTPUT 2: Eq. (13) evaluation (spatial block bootstrap by location)

def rmse(y_true, y_pred):
    y_true = np.asarray(y_true).ravel()
    y_pred = np.asarray(y_pred).ravel()
    return np.sqrt(np.mean((y_true - y_pred)**2))

def r2(y_true, y_pred):
    y_true = np.asarray(y_true).ravel()
    y_pred = np.asarray(y_pred).ravel()
    ss_res = np.sum((y_true - y_pred)**2)
    ss_tot = np.sum((y_true - np.mean(y_true))**2)
    return 1 - ss_res/ss_tot if ss_tot > 0 else np.nan

def fit_eq13_model(y, X, Wmat, shrink=0.99):
    Wmat = np.array(Wmat, dtype=float, copy=True)
    np.fill_diagonal(Wmat, 0.0)

    Ws = sp.csr_matrix(Wmat)
    w = libpysal.weights.WSP(Ws); w.transform = "R"
    w = libpysal.weights.WSP(w.sparse * shrink); w.transform = "R"

    try:
        m = ML_Lag(y=y, x=X, w=w, name_y="Y", name_x=["Y_t-1","const"])
    except Exception:
        m = GM_Lag(y=y, x=X, w=w, name_y="Y", name_x=["Y_t-1","const"])
    return m

def spatial_block_bootstrap_by_location(Y, W_ST, B=1000, seed=2026, shrink=0.99):
    rng = np.random.default_rng(seed)
    T, n = Y.shape
    N = T*n

    # build observed y and lag from Y
    Y_lag = np.zeros_like(Y)
    Y_lag[1:, :] = Y[:-1, :]
    y_obs = Y.reshape(N, 1)
    xlag_obs = Y_lag.reshape(N, 1)
    X_obs = np.hstack([xlag_obs, np.ones((N, 1))])

    # location blocks: keep all years for each location
    loc_blocks = [np.array([i + t*n for t in range(T)]) for i in range(n)]

    out = []
    failed = 0

    for b in range(1, B+1):
        chosen = rng.choice(np.arange(n), size=n, replace=True)
        idx = np.concatenate([loc_blocks[i] for i in chosen])

        y_b = y_obs[idx, :]
        X_b = X_obs[idx, :]
        W_b = W_ST[np.ix_(idx, idx)].copy()
        np.fill_diagonal(W_b, 0.0)

        try:
            model = fit_eq13_model(y_b, X_b, W_b, shrink=shrink)
            yhat = model.predy
            out.append({
                "iter": b,
                "rho": float(model.rho),
                "RMSE": float(rmse(y_b, yhat)),
                "R2": float(r2(y_b, yhat))
            })
        except Exception:
            failed += 1

    return pd.DataFrame(out), failed

boot_idw, failed = spatial_block_bootstrap_by_location(Y, W_ST_idw, B=1000, seed=2026, shrink=0.99)

print("Bootstrap success:", len(boot_idw), "failed:", failed)

R2_mean = float(boot_idw["R2"].mean())
RMSE_mean = float(boot_idw["RMSE"].mean())

print("R2_mean   :", R2_mean)
print("RMSE_mean :", RMSE_mean)

print("R2_CI95   :", (float(boot_idw["R2"].quantile(0.025)), float(boot_idw["R2"].quantile(0.975))))
print("RMSE_CI95 :", (float(boot_idw["RMSE"].quantile(0.025)), float(boot_idw["RMSE"].quantile(0.975))))
```

```python
#Export Results

summary_idw = pd.DataFrame([{
    "baseline": "Inverse Distance (row-normalized)",
    "I_obs": out_moran_idw["I_obs"],
    "I_mean": out_moran_idw["I_mean"],
    "p_value": out_moran_idw["p_value"],
    "CI95_low": out_moran_idw["CI95_low"],
    "CI95_high": out_moran_idw["CI95_high"],
    "R2_mean": R2_mean,
    "RMSE_mean": RMSE_mean
}])

summary_idw.to_csv("inverse_distance_summary_Centroid_1000rep_10th.csv", index=False)
boot_idw.to_csv("inverse_distance_bootstrap_details_Centroid_1000rep_10th.csv", index=False)

from google.colab import files
files.download("inverse_distance_summary_Centroid_1000rep_10th.csv")
files.download("inverse_distance_bootstrap_details_Centroid_1000rep_10th.csv")
```
